```python
import random

γ = 0.9
α = 0.5
ε = 0.2
states = ["s0", "s1"]
actions = ["a0", "a1"]


# transition/reward model
def step(s, a):
    if s == "s0" and a == "a0":
        return "s1", 5
    if s == "s0" and a == "a1":
        return "s0", 0
    if s == "s1" and a == "a0":
        return "s1", 2
    if s == "s1" and a == "a1":
        return "s0", 0


# epsilon‑greedy policy derived from Q
def epsilon_greedy(Q, s):
    if random.random() < ε:
        return random.choice(actions)
    # greedy tie‑break to first action
    return max(actions, key=lambda a: Q[(s, a)])


# initialise Q tables
Q_q = {(s, a): 0.0 for s in states for a in actions}
Q_sarsa = {(s, a): 0.0 for s in states for a in actions}


def run_episode(Q, on_policy=False):
    s = "s0"
    a = epsilon_greedy(Q, s) if on_policy else epsilon_greedy(Q, s)
    for _ in range(10):  # short episode
        s2, r = step(s, a)
        a2 = epsilon_greedy(Q, s2) if on_policy else epsilon_greedy(Q, s2)
        # Q‑learning target
        target_q = r + γ * max(Q[(s2, a2)] for a2 in actions)
        # SARSA target (uses the *actual* next action a2)
        target_s = r + γ * Q[(s2, a2)]
        Q_q[(s, a)] += α * (target_q - Q_q[(s, a)])
        Q_sarsa[(s, a)] += α * (target_s - Q_sarsa[(s, a)])
        s, a = s2, a2


# run many episodes
for _ in range(5000):
    run_episode(Q_q, on_policy=False)  # Q‑learning (off‑policy)
    run_episode(Q_sarsa, on_policy=True)  # SARSA (on‑policy)

print("Q‑learning:", Q_q)
print("SARSA:", Q_sarsa)
```
